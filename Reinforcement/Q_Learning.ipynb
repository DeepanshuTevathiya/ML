```python
import gymnasium as gym
import numpy as np
import random
```

```python
alpha = 0.5
gamma = 0.99
epsilon = 0.1
episodes = 500
```

```python
# Q_table
Q = np.zeros((48,4))
```

```python
# Policy
def epsilon_greedy(state):
    if random.random() < epsilon:
        return env.action_space.sample()
    else:
        return np.argmax(Q[state])
```

```python
# Q-learning

for episode in range(episodes):
    if (episode%50==0):
        env = gym.make("CliffWalking-v1", render_mode="human")
        print(f"Render__Episode: {episode+1}")
    else:
        env = gym.make("CliffWalking-v1")

    state, _ = env.reset()
    done = False

    tot_rewards = 0
    ep_length = 0

    while not done:
        action = epsilon_greedy(state)
        
        next_state, reward, terminated, truncated, _ = env.step(action)
        done = terminated or truncated

        Q[state, action] += alpha * (reward + gamma*max(Q[next_state]) - Q[state, action])

        state = next_state

        tot_rewards += reward
        ep_length += 1

    print(f"Episode: {episode+1}  |  Rewards: {tot_rewards}   |  Episode_length: {ep_length}")
    env.close()
```

```python
# What did our agent learn?
env = gym.make("CliffWalking-v1", render_mode = "human")
state, _ = env.reset()
done = False

rewards = 0
length = 0

while not done:
    action = np.argmax(Q[state])
    state, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated

    rewards += reward
    length += 1

print(f"Rewards: {rewards}  |  Length: {length}")
env.close()
```
